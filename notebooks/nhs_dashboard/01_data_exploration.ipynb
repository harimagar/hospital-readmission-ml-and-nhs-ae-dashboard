```python
import pandas as pd
#Load the CSV WITHOUT skipping anything
# This lets us inspect the raw file and find the real header row.
raw = pd.read_csv(
    r"/Users/harimagar/Documents/PROJECTS/hospital-readmission-ml-and-nhs-ae-dashboard/data/ae_raw/nhs.csv",
    header=None
)

raw.head(40)
```

```python
#Automatically detect the header row
# The real header row contains the word "Code".
# This finds the row index where "Code" appears.
header_row = raw[raw.apply(lambda row: row.astype(str).str.contains('Code').any(), axis=1)].index[0]
header_row
```

```python
# skiprows = header_row means Pandas will treat the next row as the header.
df = pd.read_csv(
    r"/Users/harimagar/Documents/PROJECTS/hospital-readmission-ml-and-nhs-ae-dashboard/data/ae_raw/nhs.csv",
    skiprows=header_row
)
```

```python
df.columns
```

```python
df = df[~((df['Code'] == '-') & (df['System'] == '-'))]
```

```python
#Clean numeric columns
# Convert numbers stored as strings with commas into floats.
# If conversion fails (e.g., text columns), we ignore it.
for col in df.columns:
    try:
        df[col] = df[col].astype(str).str.replace(',', '').astype(float)
    except:
        pass
```
